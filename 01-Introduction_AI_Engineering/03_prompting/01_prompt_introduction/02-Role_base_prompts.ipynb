### Dandole estructura a un ROl y a la selección de modelo para casos de usos particulares



```python
# Librerias necesarias para el proyecto 
import os 
from dotenv import load_dotenv
from openai import OpenAI
from enum import Enum
from IPython.display import display, Markdown
```

```python
# Cargando las variables de entorno 

load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
# Diseñando una Clase para manejar diferentes tipos de modelos 
# Nos enfocaremos en OpenAI

class OpenAIModels(str,Enum): # Str -> Para que los valores sean cadenas de texto # Enum -> Para crear una enumeración
    GPT_4o = "gpt-4o"
    GPT_4o_mini = "gpt-4o-mini"
    GPT_5_mini = "gpt-5-mini"

MODEL = OpenAIModels.GPT_5_mini # Seleccionando el modelo a utilizar
```

```python
def display_responses(*args):
    markdown_string = "<table><tr>"
    for arg in args:
        markdown_string += f"<th>System Prompt:<br />{arg['system_prompt']}<br /><br />"
        markdown_string += f"User Prompt:<br />{arg['user_prompt']}</th>"
    markdown_string += "</tr>"
    markdown_string += "<tr>"
    for arg in args:
        markdown_string += f"<td>Response:<br />{arg['response']}</td>"
    markdown_string += "</tr></table>"
    display(Markdown(markdown_string))
```

```python
def get_completion(system_prompt, # Define el comportamiento del modelo
                   user_prompt,  # Es ;la solicitud del usuario
                   model=MODEL):

    # Construcción de los mensajes 
    messages = [
        {"role": "user", "content": user_prompt},
    ]
    if system_prompt is not None:
        messages = [
            {"role": "system", "content": system_prompt},
            *messages,
        ]
    try:
        # LLamado a la API de OpenAI
        response = client.chat.completions.create(
            model=model,
            messages=messages,
            temperature=None, # Controla la creatividad de las respuestas
        )
        return response.choices[0].message.content # DEvuelve solo el cuerpo de la respuesta , no toda la metadata
    except Exception as e:
        return f"An error occurred: {e}"
```

### Ejemplo Básico

```python
control_system_prompt = "eres un asistente útil"
user_prompt = "Puedes hacer una critica sobre el libro el llamado la civilizacion del espectaculo de Vargas Llosa?"
MODEL = OpenAIModels.GPT_5_mini 
print(f"Enviando solicitud al modelo :{MODEL} ")
control_response = get_completion(control_system_prompt, user_prompt, model=MODEL)
print("Respuesta recibida!\n")

display_responses(
    {
        "system_prompt": control_system_prompt,
        "user_prompt": user_prompt,
        "response": control_response,
    }
)
```

```python
MODEL = OpenAIModels.GPT_4o # Cambiando el modelo a GPT-4o
control_system_prompt = "eres un asistente útil"
user_prompt = "Puedes hacer una critica sobre el libro el llamado la civilizacion del espectaculo de Vargas Llosa?"

print(f"Enviando solicitud al modelo :{MODEL} ")
control_response = get_completion(control_system_prompt, user_prompt, model=MODEL)
print("Respuesta recibida!\n")

display_responses(
    {
        "system_prompt": control_system_prompt,
        "user_prompt": user_prompt,
        "response": control_response,
    }
)
```

```python
# 1. Definimos el acercamiento básico (Baseline)
baseline_donjuan_prompt = "Eres Don Juan, un asistente experto en el arte de la seducción. Tu objetivo es encantar y enamorar a tu pareja ideal."
user_prompt_donjuan = "Escribe un primer mensaje breve para invitarme a tomar un café y conocernos mejor."

MODEL = OpenAIModels.GPT_5_mini 

print("Enviando solicitud inicial al modelo...")
baseline_donjuan_response = get_completion(baseline_donjuan_prompt, user_prompt_donjuan, model=MODEL)

# Evaluamos el resultado inicial
display_responses(
    {
        "system_prompt": baseline_donjuan_prompt,
        "user_prompt": user_prompt_donjuan,
        "response": baseline_donjuan_response,
    }
)
```

```python
# 2. Le damos profundidad al rol
persona_donjuan_prompt = f"""
{baseline_donjuan_prompt}

Adopta las siguientes características para tu personalidad:
1. Usa la poesía clásica española como inspiración (simulando a autores como Bécquer o Lope de Vega).
2. Tienes una confianza inquebrantable, pero siempre mantienes un respeto absoluto y caballeroso.
3. Haz observaciones detalladas sobre la belleza de los pequeños detalles (una mirada, una sonrisa sincera).
4. Utiliza metáforas relacionadas con el mar, las estrellas y el destino.
"""

print("Enviando solicitud con atributos de personalidad...")
persona_donjuan_response = get_completion(persona_donjuan_prompt, user_prompt_donjuan)

# Comparamos la evolución
display_responses(
    {
        "system_prompt": baseline_donjuan_prompt,
        "user_prompt": user_prompt_donjuan,
        "response": baseline_donjuan_response,
    },
    {
        "system_prompt": persona_donjuan_prompt,
        "user_prompt": user_prompt_donjuan,
        "response": persona_donjuan_response,
    }
)
```

```python
# 3. Afinamos el Tono, Estilo e imponemos Restricciones
tone_donjuan_prompt = f"""
{persona_donjuan_prompt}

Tono y Estilo:
- Usa expresiones de asombro sutil, como: "Dime, hermosa criatura..." o "Es curioso cómo..."
- Mantén un tono de voz bajo, íntimo y misterioso (como si le estuvieras susurrando al oído).
- Intercala algún halago coqueto pero muy elegante.

Restricciones:
- No escribas más de 3 o 4 oraciones. Tienes que ser directo pero impactante.
- No uses emojis vulgares, solo si acaso una 🌹.
"""

print("Enviando solicitud final con Tono y Estilo...")
tone_donjuan_response = get_completion(tone_donjuan_prompt, user_prompt_donjuan)

# Comparamos el Baseline inicial vs la versión final (Evolución completa)
display_responses(
    {
        "system_prompt": baseline_donjuan_prompt,
        "user_prompt": user_prompt_donjuan,
        "response": baseline_donjuan_response,
    },
    {
        "system_prompt": tone_donjuan_prompt,
        "user_prompt": user_prompt_donjuan,
        "response": tone_donjuan_response,
    }
)
```
